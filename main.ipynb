```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split,cross_val_score # split the data set for train as test.
from sklearn.preprocessing import StandardScaler,OneHotEncoder 
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
#Import data from dataset.
df = pd.read_csv('CarsData.csv') # Paste here your dataset path. 
df
```

```python
#Infrormation about DataFrame.
df.info()
```

```python
#Find the Null sum of each columns.
df.isnull().sum()
```

```python
# Copy the dataFrame into another Frame.
sub_df = df.copy()
```

```python
target_set = sub_df['price'].copy() # separate the Target set from dataSet
sub_df = sub_df.drop(['price'],axis='columns')
```

```python
target_set
```

```python
sub_df
```

```python
num_clms = sub_df.drop(['model','transmission','fuelType','Manufacturer'],axis='columns').columns.tolist()
category_clms = ['model','transmission','fuelType','Manufacturer']
```

```python
num_clms
```

```python
category_clms
```

```python
#makeing a pipeline for Numerical columns .
num_pipeline = Pipeline([
    ('Imputer',SimpleImputer(strategy="median")),
    ('Scaling',StandardScaler())
])
```

```python
#Makeing a pipeline for categorical columns.
category_pipeline = Pipeline([
    ('ohe',OneHotEncoder(handle_unknown="ignore"))
])
```

```python
#constructing a Full pipeline.
full_pipeline = ColumnTransformer([
    ('num_pipeline',num_pipeline,num_clms),
    ('category_pip',category_pipeline,category_clms)
])
```

```python
x_train = full_pipeline.fit_transform(sub_df)
```

```python
#---------------------------------------- MODEL TRAINING --------------------------------------------------
```

```python
# Train a model using Linear Regression .
lin_reg = LinearRegression()
lin_reg_score = cross_val_score(lin_reg, x_train, target_set , scoring='r2', cv = 10)
```

```python
print("R2 scores : ",lin_reg_score) 
print("Mean of R2 score : ",lin_reg_score.mean())
```

```python
# Train model using DecisionTreeRegression.
dict_reg = DecisionTreeRegressor(max_depth=20, min_samples_leaf=5, random_state= 42 )
dict_reg_score = cross_val_score(dict_reg, x_train, target_set , scoring='r2', cv = 10)
```

```python
#Score of RandomForestRegression.
print("R2 scores : ",dict_reg_score) 
print("Mean of R2 score : ",dict_reg_score.mean())
```

```python
# Train a model using RandomForestRegression.
rdmf_reg = RandomForestRegressor(n_estimators=200, max_depth=20, min_samples_leaf=5, n_jobs=-1, random_state=42)
rdmf_reg_score = cross_val_score(rdmf_reg, x_train, target_set , scoring='r2', cv = 10)
```

```python
#Score of RandomForestRegression.
print("R2 score : ",rdmf_reg_score)
print("Mean of R2 score : ",rdmf_reg_score.mean())
```

```python
# 
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=8,
    random_state=42
)

xgb_pipeline = Pipeline([
    ("preprocessing",full_pipeline),
    ('model',xgb_model)
])
```

```python
xgb_score = cross_val_score(xgb_pipeline, sub_df, target_set , scoring='r2', cv = 5)
```

```python
#Score of XGBRegressor.
print("R2 score : ",xgb_score)
print("Mean of R2 score : ",xgb_score.mean())
```
